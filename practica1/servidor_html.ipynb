## Objetivo general:

- Comprender los campos que conforman la trama Ethernet 802.3

## Objetivos específicos:

- Comprender como analizar tramas con Wireshark

## Requisitos:

Instale Wireshark y Python de ser necesario, en su computador.

# Levantar un servidor web simple usando Python

El objetivo de esta guía es **levantar un servidor web simple usando Python**, sirviendo un archivo `index.html`.

---
## 1. Crear estructura de archivos
```
mi-web/
 ├── index.html
```

---
## 2. Crear `index.html`
Personalizar el archivo para que muestre su nombre.

```html
<!DOCTYPE html>
<html>
<head>
    <meta charset="UTF-8">
    <title>Mi Página</title>
</head>
<body>
    <h1>Hola, soy Pepe</h1>
</body>
</html>
```

---
## 3. Ejecutar el servidor
```bash
python -m http.server 80
```

### Nota:
- Verificar que el puerto 80 no esté siendo usado por otro servidor.
- En caso de que no pueda levantar el servidor cambie el puerto por 8080,  8081, etc.

# Crear la tabla de IPs

El objetivo en este punto es crear una tabla con el nombre, IP y MAC de las computadoras de sus compañeros. Para conseguirlo deberá usar las herramientas propuestas.

## 1. Rastrear las IPs de las PCs dentro de la misma red

Debe obtener las direcciones IP de las computadoras de sus compañeros, esto es posible si todas están dentro de la misma red, por lo que el uso de un router exclusivo será necesario.

## 2. Verificar la pertenencia de la IP

Abra la web de las IPs que encontró y registre en la tabla a quién pertenece la IP.

## 3. Verificar las tramas e identificar la MAC

En paralelo ejecute Wireshark y analice las tramas para identificar la MAC de origen y destino.

## 4. Comprobar que los datos estén correctos

En este punto se harán públicos los datos de IP y MAC de las computadoras de sus compañeros para validar el experimento.
